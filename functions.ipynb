```python
# 4. funkcje z wartością zwracaną (jedna i wiele)
def add(a, b):
    return a + b

result = add(5, 3)
print(result)

def calculate(a, b):
    sum_ = a + b
    product = a * b
    return sum_, product

sum_result, product_result = calculate(5, 3)
print(f"Suma: {sum_result}, Iloczyn: {product_result}")
```

```python
# 7. funkcje z rekurencją
def silnia(n):
    if n == 0:
        return 1
    else:
        return n * silnia(n-1)
    
print(silnia(5))  # Output: 120
```

Ćwiczenie 3. Stwórz funkcje rekurencyjną która obliczy sume liczb od 0 do 10


```python
# 8. variable shadowing (global)
x = 10

def shadowing_example():
    x = 5  # To jest zmienna lokalna, która "zasłania" globalną zmienną x
    print(f"Wartość x wewnątrz funkcji: {x}")
    
def shadowing_with_global():
    global x  # Deklarujemy, że chcemy użyć globalnej zmiennej x
    x = 20  # Modyfikujemy globalną zmienną x
    print(f"Wartość x wewnątrz funkcji (po modyfikacji): {x}")

shadowing_example()
print(f"Wartość x poza funkcją: {x}")

shadowing_with_global()
print(f"Wartość x poza funkcją (po modyfikacji): {x}")
```

```python
# 9. variable shadowing (nonlocal)
def outer_function():
    x = 5
    def inner_function():
        nonlocal x
        x = 10
        print("Inner:", x)
    inner_function()
    print("Outer:", x)
outer_function()
```

```python
# 10. generatory
def generuj_liczby():
    for i in range():
        yield i * 2

for liczba in generuj_liczby():
    print(liczba)
```
