```python
# 1. try except (np. ZeroDivisionError)
try:
    a = 10 / 0
except ZeroDivisionError:
    print("Nie można dzielić przez 0!")
    
# as e other exceptions
try:
    a = 10 / 0
except ZeroDivisionError as e:
    print(f"Error: {e}")
```

```python
# 2. try except except
# ValueError
try:
    num = int(input("Wprowadź liczbę: "))
    a = 10 / num
except ValueError:
    print("To nie jest prawidłowa liczba!")
except ZeroDivisionError:
    print("Nie można dzielić przez 0!")
except Exception as e:
    print(f"Wystąpił nieoczekiwany błąd: {e}")
```

```python
# 3. try except finally
try:
    a = 10 / 0
except ZeroDivisionError:
    print("Nie można dzielić przez 0!")
finally:
    print("To zawsze zostanie wykonane.")
```

```python
# 4. ponowne rzucanie wyjątku
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Nie można dzielić przez 0!")
        raise  # re-raise the exception to be handled by an outer block
    
try:
    result = divide(10, 0)
except ZeroDivisionError:
    print("Obsłużono wyjątek w zewnętrznym bloku.")

```
